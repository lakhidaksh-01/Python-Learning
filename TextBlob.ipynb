TextBlob Cheat Sheet super useful for sentiment analysis

```python
from textblob import TextBlob
```

```python
blob = TextBlob("I love Python. It's amazing!")
```

```python
blob.sentiment            # Sentiment(polarity=0.75, subjectivity=0.8)
blob.sentiment.polarity   # -1 = bad, 0 = neutral, +1 = good
blob.sentiment.subjectivity  # 0 = fact, 1 = opinion
```

```python
blob.words       # ['I', 'love', 'Python', 'It', 'is', 'amazing']
blob.sentences   # [Sentence("I love Python."), Sentence("It's amazing!")]
```

```python
TextBlob("I havv a speling errrorr").correct()
# "I have a spelling error"

word = TextBlob("studies").words[0]
word.singularize()   # study
word.pluralize()     # studies → studies (depends on word)
```

```python
import pandas as pd

df = pd.DataFrame({'comment': ["I love it", "This is bad", "Not bad"]})

df['sentiment'] = df['comment'].apply(lambda x: TextBlob(x).sentiment.polarity)

print(df)
```
